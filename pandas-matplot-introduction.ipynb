#### 8. Matplotlib for visulizing data
goal: what is matplotlib and why is etical for ML/AI



```python
import matplotlib.pyplot as plt
```

```python
# read titanic dataframe get head and shape of it
titanic = pd.read_csv("titanic.csv")
```

#### 9. Basic Plots: Line and Scatter

```python
ages = titanic['Age'].dropna()

plt.figure(figsize=(8, 5))  
plt.plot(ages.index[:50], ages[:50], color='blue', marker='o')  
plt.title("First 50 Passenger Age") 
plt.xlabel("Passenger Index")  
plt.ylabel('Age')  
plt.show()
```

AI/ML Tip: Scatter helps you see correlation (like age and fare) for feature selection in a regression model.

```python
plt.figure(figsize=(8, 5))
plt.scatter(titanic['Age'], titanic['Fare'], color='red', alpha=0.6)  
plt.title("Distrubtion age vs fare")
plt.xlabel('Age')
plt.ylabel("Fare")
plt.show()
```

#### 10. Histograms And Density Plots
goal: See the distribution of data.

```python
plt.figure(figsize=(8, 5))
plt.hist(titanic['Age'].dropna(), bins=20, color='green', edgecolor='black') 
plt.title("Histogram Age of Passengers")
plt.xlabel('Age')
plt.ylabel("Number")
plt.show()
```

```python
titanic['Age'].hist(density=True, bins=20, color='orange')  
plt.title("Age Density")
plt.show()
```

#### 11. Bar And Pie Charts
goal: Group comparison (e.g. survival by sex).

```python
survival_by_sex = titanic.groupby('Sex')['Survived'].mean() * 100  
print(survival_by_sex)  # female: ~74%, male: ~19%

plt.figure(figsize=(6, 4))
survival_by_sex.plot(kind='bar', color=['pink', 'lightblue'])
plt.title("Survived based on gender")
plt.xlabel('Gender')
plt.ylabel("Survive Precent")
plt.show()
```

```python
class_counts = titanic['Pclass'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(class_counts.values, labels=class_counts.index, autopct='%1.1f%%')
plt.title("Tiket class Distrbuation")
plt.show()
```

#### 12. Subplots and chart combinations
goal: plot multiple chart in one figure

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5)) 

# subplot 1: Age Histogram
ax1.hist(titanic['Age'].dropna(), bins=15, color='blue')
ax1.set_title("Age Density")
ax1.set_xlabel('Age')
ax1.set_ylabel('Number')

# subplot ۲: Bar بقا
survival_by_class = titanic.groupby('Pclass')['Survived'].mean()
ax2.bar(survival_by_class.index, survival_by_class.values, color='orange')
ax2.set_title('Average survival by class')
ax2.set_xlabel('Class')
ax2.set_ylabel('Average survival')

plt.tight_layout() 
plt.show()
```
